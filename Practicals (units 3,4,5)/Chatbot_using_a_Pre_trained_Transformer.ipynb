```python
!pip -q install transformers
from transformers import pipeline, set_seed

# Load the pretrained text-generation model
chatbot = pipeline(
    "text-generation",
    model="distilgpt2"
)

set_seed(42)

print("Transformer Chatbot")
print("Type 'exit' to stop.\n")

while True:
    user_input = input("You: ").strip()

    if user_input.lower() == "exit":
        print("Bot: Goodbye!")
        break

    if not user_input:
        print("Bot: Please enter a message.")
        continue

    prompt = (
        "The following is a conversation.\n"
        f"User: {user_input}\n"
        "Bot:"
    )

    response = chatbot(
        prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.8,
        pad_token_id=chatbot.tokenizer.eos_token_id
    )[0]["generated_text"]

    reply = response[len(prompt):].strip()

    if not reply:
        reply = "I am still learning. Please try another question."

    print("Bot:", reply, "\n")
```
