## GloVe Embeddings

GloVe (Global Vectors for Word Representation) is an unsupervised learning algorithm for obtaining vector representations for words. It learns word embeddings by factoring a word-word co-occurrence matrix, capturing both global statistics and local context. These embeddings are widely used in natural language processing (NLP) tasks for representing words as dense vectors, which can then be used as input for machine learning models.

```python
import numpy as np

def load_glove_embeddings(filepath):
    """Loads GloVe embeddings from a specified file."""
    embeddings_index = {}
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            coefs = np.asarray(values[1:], dtype='float32')
            embeddings_index[word] = coefs
    return embeddings_index

# Load the 100-dimensional GloVe embeddings
glove_embeddings = load_glove_embeddings('glove.6B.100d.txt')

print(f"Loaded {len(glove_embeddings)} word vectors.")
print(f"Vector dimension: {len(next(iter(glove_embeddings.values())))}")
```

### Word Vector Representation

We can retrieve the vector for a single word by looking it up in the loaded `glove_embeddings` dictionary.

```python
def get_word_vector(word, embeddings):
    """Returns the GloVe vector for a given word, or a zero vector if not found."""
    return embeddings.get(word.lower(), np.zeros(100))

# Example usage for a single word
word = "apple"
word_vector = get_word_vector(word, glove_embeddings)
print(f"Vector for '{word}':\n{word_vector[:5]}... (first 5 dimensions)") # Print first 5 dimensions for brevity
print(f"Vector shape: {word_vector.shape}")

word_not_found = "nonexistentword123"
word_vector_not_found = get_word_vector(word_not_found, glove_embeddings)
print(f"\nVector for '{word_not_found}' (not found):\n{word_vector_not_found[:5]}... (first 5 dimensions)")
```

### Sentence Vector Representation (Averaging Word Vectors)

One common way to represent a sentence as a single vector is by averaging the vectors of all the words within the sentence. This method provides a simple yet effective baseline for many NLP tasks.

```python
def get_sentence_vector(sentence, embeddings, vector_dim=100):
    """Averages word vectors to create a sentence vector."""
    words = sentence.lower().split()
    # Filter out words not in embeddings to avoid zero vectors from influencing average if a word is truly unknown
    valid_word_vectors = [embeddings[word] for word in words if word in embeddings]

    if not valid_word_vectors:
        return np.zeros(vector_dim) # Return a zero vector if no words are found

    return np.mean(valid_word_vectors, axis=0)

# Example usage for a sentence
sentence = "This is a sample sentence about natural language processing."
sentence_vector = get_sentence_vector(sentence, glove_embeddings)
print(f"Sentence: '{sentence}'")
print(f"Sentence vector (first 5 dimensions):\n{sentence_vector[:5]}...")
print(f"Sentence vector shape: {sentence_vector.shape}")
```
