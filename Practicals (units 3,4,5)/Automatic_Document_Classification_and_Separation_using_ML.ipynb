```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Sample training documents
documents = [
    "Python programming and artificial intelligence",
    "Machine learning and neural networks",
    "Football match and cricket tournament",
    "Players scored goals in the match",
    "Healthy food and regular exercise",
    "Doctors provide medical treatment"
]

labels = [
    "Technology",
    "Technology",
    "Sports",
    "Sports",
    "Health",
    "Health"
]

# Build the classification pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", MultinomialNB())
])

# Train the classifier
model.fit(documents, labels)

# Classify new documents
new_documents = [
    "Deep learning and computer programming",
    "The team won the football match",
    "Exercise improves physical health"
]

predictions = model.predict(new_documents)

print("Document Classification Results:\n")

for document, category in zip(new_documents, predictions):
    print("Document:", document)
    print("Category:", category)
    print("-" * 50)

# Separate documents according to their predicted category
separated_documents = {}

for document, category in zip(new_documents, predictions):
    separated_documents.setdefault(category, []).append(document)

print("\nSeparated Documents:")

for category, docs in separated_documents.items():
    print(f"\n{category}:")
    for doc in docs:
        print("-", doc)
```
