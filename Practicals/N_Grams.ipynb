```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt')
except LookupError: # Corrected exception type
    nltk.download('punkt')

def generate_unigrams(texts):
    """
    Generates unigrams (single words) from a given list of texts (documents).
    """
    all_unigrams = []
    for text in texts:
        # Tokenize the text into words
        words = word_tokenize(text.lower()) # Convert to lowercase for consistency
        all_unigrams.extend(words)
    return all_unigrams
```

```python
# Define your corpus here as a list of documents
corpus = [
    "This is a sample document.",
    "It can contain multiple sentences.",
    "Each sentence is considered a document in this corpus."
]
```

```python
# Generate unigrams from the corpus
unigrams = generate_unigrams(corpus)

# Create a DataFrame
unigram_df = pd.DataFrame({"Unigram": unigrams})

# Display the DataFrame
display(unigram_df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Fit and transform the corpus to create the document-term matrix
document_term_matrix = vectorizer.fit_transform(corpus)

# Get the feature names (unigrams)
feature_names = vectorizer.get_feature_names_out()

# Create a DataFrame for better readability
dtm_df = pd.DataFrame(document_term_matrix.toarray(), columns=feature_names)

# Display the document-term matrix
print("Document-Term Matrix:")
display(dtm_df)
```
