# GloVe Embeddings Loading and Vector Representation
This notebook demonstrates how to load pre-trained GloVe (Global Vectors for Word Representation) embeddings and use them to construct vector representations for words and sentences.

```python
!pip install gensim

import gensim.downloader as api
import numpy as np

# Load pre-trained GloVe embeddings (100 dimensions)
# This might take a minute to download on the first run
print("Loading GloVe embeddings...")
glove_vectors = api.load("glove-wiki-gigaword-100")
print("Embeddings loaded successfully!")
```

## Exploring the Embeddings
Let's look up individual word vectors and find similar words to verify the embeddings loaded correctly.

```python
# Get vector for a single word
word = "coffee"
vector = glove_vectors[word]
print(f"Vector for '{word}' (first 10 dimensions):\n{vector[:10]}\n")

# Find top 5 most similar words
similar_words = glove_vectors.most_similar(word, topn=5)
print(f"Words most similar to '{word}':")
for w, similarity in similar_words:
    print(f"  {w}: {similarity:.4f}")
```

## Sentence Representation (Vector Averaging)
To represent a whole sentence as a single vector, a common baseline approach is to average the GloVe vectors of all the individual words present in our vocabulary.

```python
def get_sentence_vector(sentence, model):
    # Tokenize sentence and lowercase
    words = sentence.lower().split()
    # Filter out words that are not in the vocabulary
    valid_vectors = [model[w] for w in words if w in model]

    if not valid_vectors:
        # Return a zero vector if no words are in vocabulary
        return np.zeros(model.vector_size)

    # Average the vectors
    return np.mean(valid_vectors, axis=0)

# Example usage
sentence_1 = "Data science is exciting"
sentence_2 = "Machine learning is fun"

vec_1 = get_sentence_vector(sentence_1, glove_vectors)
vec_2 = get_sentence_vector(sentence_2, glove_vectors)

# Calculate cosine similarity
cosine_similarity = np.dot(vec_1, vec_2) / (np.linalg.norm(vec_1) * np.linalg.norm(vec_2))
print(f"Sentence 1: '{sentence_1}'")
print(f"Sentence 2: '{sentence_2}'")
print(f"Cosine Similarity: {cosine_similarity:.4f}")
```
