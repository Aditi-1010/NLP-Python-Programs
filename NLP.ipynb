### Comparing Lemmatization with NLTK and SpaCy

Let's use the same text and apply lemmatization with both NLTK's `WordNetLemmatizer` and SpaCy to observe the differences.

```python
import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab') # Added to resolve LookupError
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

import spacy

# Sample text for comparison
text_to_compare = "The cats were running and jumping happily. They have eaten all the mice."

print(f"Original Text: {text_to_compare}\n")

# --- NLTK Lemmatization ---
lemmatizer_nltk = WordNetLemmatizer()
words_nltk = word_tokenize(text_to_compare)

lemmatized_words_nltk = []
for word in words_nltk:
    # NLTK's lemmatizer benefits from POS tagging for better accuracy.
    # Without POS, it defaults to noun. We'll show a simple version first.
    lemmatized_words_nltk.append(lemmatizer_nltk.lemmatize(word.lower()))

print("NLTK Lemmatization (without POS tag):")
print(lemmatized_words_nltk)
print("\n")

# --- SpaCy Lemmatization ---
nlp_spacy = spacy.load("en_core_web_sm")
doc_spacy = nlp_spacy(text_to_compare)

lemmatized_words_spacy = [token.lemma_ for token in doc_spacy]

print("SpaCy Lemmatization:")
print(lemmatized_words_spacy)
```

### Explanation of Differences

You'll likely observe differences due to how each library handles lemmatization:

*   **NLTK's `WordNetLemmatizer`**: By default, it assumes words are nouns if no Part-of-Speech (POS) tag is provided. For better accuracy, you need to explicitly pass the POS tag (e.g., `lemmatizer.lemmatize(word, pos='v')` for verbs). This makes NLTK more flexible but requires more manual control.
*   **SpaCy**: SpaCy's `nlp` pipeline includes a POS tagger and dependency parser by default. It uses this contextual information to determine the correct lemma for each word, often leading to more accurate results out-of-the-box without explicit POS tagging.

### Stop Word Removal without NLTK (Manual Approach)

In this approach, we define a custom list of common stop words and filter them out from our text tokens.

```python

text_for_stopwords = "This is a very good example of text that contains common stop words."

manual_stopwords = ["a", "an", "the", "is", "are", "was", "were", "be", "been", "being", "of", "to", "in", "on", "at", "for", "with", "and", "but", "or", "this", "that", "these", "those", "i", "me", "my", "you", "your", "he", "him", "his", "she", "her", "its", "we", "us", "our", "they", "them", "their", "it", "which", "what", "when", "where", "why", "how"]

words = text_for_stopwords.lower().split()

filtered_words_manual = [word for word in words if word not in manual_stopwords]

print(f"Original Text: {text_for_stopwords}")
print(f"Tokens: {words}")
print(f"After Manual Stop Word Removal: {filtered_words_manual}")
```

### Stop Word Removal with NLTK

NLTK provides a comprehensive list of stop words for various languages, making it easy to perform this task.

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text_for_stopwords_nltk = "This is a very good example of text that contains common stop words."

stop_words_nltk = set(stopwords.words('english'))

words_nltk = word_tokenize(text_for_stopwords_nltk.lower())

filtered_words_nltk = [word for word in words_nltk if word not in stop_words_nltk]

print(f"Original Text: {text_for_stopwords_nltk}")
print(f"Tokens: {words_nltk}")
print(f"After NLTK Stop Word Removal: {filtered_words_nltk}")
```

### Part-of-Speech (POS) Tagging

POS tagging is the process of marking up a word in a text (corpus) as corresponding to a particular part of speech, based on both its definition and its context. Common parts of speech include nouns, verbs, adjectives, adverbs, pronouns, prepositions, conjunctions, interjections, and determiners.

Let's see how NLTK and SpaCy perform POS tagging.

```python
import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng') # Added to resolve LookupError
nltk.download('punkt') # Ensure tokenizer is available
from nltk.tokenize import word_tokenize

# Sample text
pos_text = "The quick brown fox jumps over the lazy dog."

print(f"Original Text: {pos_text}\n")

# Tokenize the text
words = word_tokenize(pos_text)

# Perform POS tagging with NLTK
pos_tags_nltk = nltk.pos_tag(words)

print("NLTK POS Tagging:")
for word, tag in pos_tags_nltk:
    print(f"{word}: {tag}")
```

### SpaCy POS Tagging

SpaCy's models often provide more detailed and context-aware POS tags, along with other linguistic annotations like dependency parsing.

```python
import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")

# Sample text
pos_text = "The quick brown fox jumps over the lazy dog."

# Process the text with SpaCy
doc = nlp(pos_text)

print("SpaCy POS Tagging:")
for token in doc:
    # token.text: the original word
    # token.pos_: the coarse-grained part-of-speech tag (e.g., NOUN, VERB)
    # token.tag_: the fine-grained part-of-speech tag (e.g., NNS, VBP)
    print(f"{token.text}: {token.pos_} ({token.tag_})")
```

## Hidden Markov Models (HMMs)

A **Hidden Markov Model (HMM)** is a statistical Markov model in which the system being modeled is assumed to be a Markov process with unobserved (hidden) states. An HMM is characterized by:

1.  **Hidden States (Q)**: A set of states that are not directly observable. These states follow a Markov chain, meaning the probability of transitioning to the next state depends only on the current state.
2.  **Observation Symbols (V)**: A set of observable outputs that are generated by the hidden states.
3.  **Initial State Distribution (π)**: The probability distribution over the initial hidden states.
4.  **Transition Probability Matrix (A)**: The probabilities of moving from one hidden state to another.
5.  **Emission (or Observation) Probability Matrix (B)**: The probabilities of observing a particular symbol from a given hidden state.

In NLP, HMMs are often used for tasks like Part-of-Speech (POS) tagging, speech recognition, and machine translation. For example, in POS tagging, the hidden states could be the grammatical tags (e.g., Noun, Verb, Adjective), and the observations would be the words in the sentence. The HMM tries to find the most likely sequence of hidden states (POS tags) given the observed sequence of words.

### Key Problems Solved by HMMs:

*   **Evaluation Problem**: Given an HMM and a sequence of observations, what is the probability that the model generated this sequence? (Solved by the Forward Algorithm).
*   **Decoding Problem**: Given an HMM and a sequence of observations, what is the most likely sequence of hidden states that generated these observations? (Solved by the Viterbi Algorithm).
*   **Learning Problem**: Given a set of observation sequences, how can we learn the HMM parameters (A, B, π)? (Solved by the Baum-Welch Algorithm).

Now, let's create a simple HMM example. Imagine a situation where you are in a hidden state (sunny or rainy) and you observe activities (walking, shopping, cleaning). We'll train an HMM to learn the patterns.

```python
import numpy as np
from hmmlearn import hmm

# Define the hidden states: 0 = 'Sunny', 1 = 'Rainy'
# Define the observations: 0 = 'Walk', 1 = 'Shop', 2 = 'Clean'

# --- 1. Initial State Distribution (π) ---
# Let's assume we start sunny 60% of the time, rainy 40% of the time
start_probability = np.array([0.6, 0.4])

# --- 2. Transition Probability Matrix (A) ---
# Transition probabilities from state i to state j
# A[i, j] = P(state j at t+1 | state i at t)
# E.g., if Sunny today, 70% chance Sunny tomorrow, 30% chance Rainy tomorrow
# If Rainy today, 40% chance Sunny tomorrow, 60% chance Rainy tomorrow
transition_probability = np.array([
    [0.7, 0.3],  # Sunny -> Sunny, Sunny -> Rainy
    [0.4, 0.6]   # Rainy -> Sunny, Rainy -> Rainy
])

# --- 3. Emission (Observation) Probability Matrix (B) ---
# Emission probabilities from state i observing symbol j
# B[i, j] = P(observation j | state i)
# E.g., if Sunny, 10% clean, 50% walk, 40% shop
# If Rainy, 60% clean, 10% walk, 30% shop
emission_probability = np.array([
    [0.1, 0.5, 0.4],  # Sunny: Clean, Walk, Shop
    [0.6, 0.1, 0.3]   # Rainy: Clean, Walk, Shop
])

# Create an HMM model with 2 hidden states
model = hmm.MultinomialHMM(n_components=2)

# Assign the manually defined parameters to the model
model.startprob_ = start_probability
model.transmat_ = transition_probability
model.emissionprob_ = emission_probability

print("HMM Model Parameters (manually defined):")
print("Start Probabilities:\n", model.startprob_)
print("Transition Matrix:\n", model.transmat_)
print("Emission Matrix:\n", model.emissionprob_)

# --- Decoding Problem: Given observations, find the most likely sequence of hidden states ---
# Observation sequence: Walk, Shop, Walk, Clean, Shop
# Corresponds to: 1, 2, 1, 0, 2
observations = np.array([[1], [2], [1], [0], [2]])

# Predict the hidden states using the Viterbi algorithm
logprob, hidden_states = model.decode(observations, algorithm="viterbi")

print("\nObservation Sequence (0=Clean, 1=Walk, 2=Shop):\n", observations.flatten())
print("Log Probability of sequence:", logprob)
print("Most likely hidden states (0=Sunny, 1=Rainy):\n", hidden_states)

# Map states to names for better readability
state_map = {0: 'Sunny', 1: 'Rainy'}
observed_map = {0: 'Clean', 1: 'Walk', 2: 'Shop'}

print("\nDecoded Sequence:")
for i, obs_idx in enumerate(observations.flatten()):
    print(f"Day {i+1}: Observed '{observed_map[obs_idx]}' -> Most Likely State '{state_map[hidden_states[i]]}'")
```
