```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from collections import Counter
import string

# text
text_for_bow = "The quick brown fox jumps over the lazy dog. The dog is brown and quick."
print(f"Original Text: {text_for_bow}\n")

# 1. Tokenization and punctuation removal
words = [word for word in word_tokenize(text_for_bow.lower()) if word not in string.punctuation]
print(f"Tokens (lowercased, no punctuation): {words}\n")

# 2. Create the Bag of Words (word frequencies)
bow = Counter(words)
print("Bag of Words:")
for word, count in bow.items():
    print(f"'{word}': {count}")
```

### Using a Corpus to Create a Bag of Words Matrix

Let's extend the concept to multiple documents, forming a corpus. We'll create a Bag of Words matrix where each row represents a document and each column represents a unique word from the entire corpus.

```python
import pandas as pd
from nltk.tokenize import word_tokenize
from collections import Counter
import string

# Define a sample corpus (list of documents)
corpus = [
    "The quick brown fox jumps over the lazy dog.",
    "The dog is brown and quick.",
    "A quick brown cat sleeps soundly.",
    "The fox and the cat are friends."
]

# Prepare a list to store Bag of Words for each document
document_bows = []

# Process each document in the corpus
for i, doc in enumerate(corpus):
    # 1. Tokenization and punctuation removal
    words = [word for word in word_tokenize(doc.lower()) if word not in string.punctuation]

    # 2. Create the Bag of Words (word frequencies) for the current document
    bow = Counter(words)
    document_bows.append(bow)

    print(f"Document {i+1} Original Text: {doc}")
    print(f"Document {i+1} Tokens (lowercased, no punctuation): {words}")
    print(f"Document {i+1} Bag of Words: {dict(bow)}\n")

# 3. Create the Bag of Words Matrix for the entire corpus
# Fill missing words with 0 (for words not present in a specific document)
bow_corpus_matrix = pd.DataFrame(document_bows).fillna(0).astype(int)

print("Corpus Bag of Words Matrix (counts):")
display(bow_corpus_matrix)
```

```python
# 4. Create a Binary Bag of Words Matrix for the entire corpus
# Convert counts to 0s and 1s
binary_bow_corpus_matrix = (bow_corpus_matrix > 0).astype(int)

print("Corpus Binary Bag of Words Matrix (0s and 1s):")
display(binary_bow_corpus_matrix)
```
