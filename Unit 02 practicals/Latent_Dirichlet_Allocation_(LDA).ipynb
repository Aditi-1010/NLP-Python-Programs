```python
import nltk
import gensim
from gensim import corpora
from nltk.corpus import stopwords

nltk.download('stopwords')

documents = [
    "machine learning and artificial intelligence are growing",
    "deep learning uses neural networks for artificial intelligence",
    "football players score goals in the match",
    "cricket players hit runs and wickets",
    "machine learning models learn from data",
    "football and cricket are popular sports"
]

stop_words = set(stopwords.words('english'))

# Tokenization and preprocessing
processed_docs = []

for doc in documents:
    words = doc.lower().split()
    words = [word for word in words if word not in stop_words]
    processed_docs.append(words)

# Dictionary
dictionary = corpora.Dictionary(processed_docs)

# Bag of Words
corpus = [dictionary.doc2bow(doc) for doc in processed_docs]

# LDA Model
lda_model = gensim.models.LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=2,
    random_state=42,
    passes=10
)

# Display topics
print("Topics:\n")

for topic in lda_model.print_topics(num_words=5):
    print(topic)
```
