```python
!pip install -q --upgrade gensim

import urllib.request
import zipfile
import os
from gensim.models import Word2Vec
from gensim.models.word2vec import Text8Corpus

# Manually download and extract the text8 dataset to bypass gensim's buggy downloader
dataset_url = "http://mattmahoney.net/dc/text8.zip"
zip_path = "text8.zip"
corpus_path = "text8"

if not os.path.exists(corpus_path):
    print("Downloading text8 dataset manually...")
    urllib.request.urlretrieve(dataset_url, zip_path)
    print("Extracting text8 dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall()
    print("Extraction complete!")

# Load dataset using Gensim's native Text8Corpus class
print("Loading dataset corpus...")
dataset = Text8Corpus(corpus_path)

# Train a basic Word2Vec model
print("Training Word2Vec model...")
model = Word2Vec(sentences=dataset, vector_size=100, window=5, min_count=150, workers=4)

# Save the model
model.save("word2vec_text8.model")
print("Model trained and saved successfully!")

# Demonstrate basic operations
word_to_test = "king"
if word_to_test in model.wv:
    print(f"\nWords most similar to '{word_to_test}':")
    similar_words = model.wv.most_similar(word_to_test, topn=5)
    for word, similarity in similar_words:
        print(f"  {word}: {similarity:.4f}")
```
