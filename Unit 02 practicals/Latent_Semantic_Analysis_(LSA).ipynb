```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

documents = [
    "machine learning is a part of artificial intelligence",
    "deep learning uses neural networks",
    "artificial intelligence is changing technology",
    "football is a popular sport",
    "cricket players score runs",
    "football players score goals"
]

# TF-IDF
vectorizer = TfidfVectorizer(stop_words='english')

X = vectorizer.fit_transform(documents)

# LSA using SVD
lsa = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_lsa = lsa.fit_transform(X)

print("LSA Matrix:")
print(X_lsa)

# Display important words in each topic
terms = vectorizer.get_feature_names_out()

for i, component in enumerate(lsa.components_):
    print("\nTopic", i + 1)

    important_words = [
        terms[index]
        for index in component.argsort()[-5:][::-1]
    ]

    print(important_words)
```
