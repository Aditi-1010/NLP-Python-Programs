```python
#Tokenization using NLTK

import nltk
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError
from nltk.tokenize import sent_tokenize, word_tokenize
text_nltk = "Natural Language Processing is interesting. It is used in Artificial Intelligence" # Renamed text to avoid conflict
sentences = sent_tokenize(text_nltk)
for sentence in sentences:
  print(sentence)

# For word tokenization with NLTK, we need to define 'tokens' explicitly.
tokens_nltk = word_tokenize(text_nltk)
for token in tokens_nltk:
  print(token)

# Tokenization using spaCy

import spacy
nlp = spacy.load("en_core_web_sm")
text_spacy = "Natural Language Processing is a part of AI. It helps computers understand human language." # Renamed text to avoid conflict
doc = nlp(text_spacy)
print("\nSentence Tokenization (spaCy):")
for sent in doc.sents:
    print(sent.text)
print("\nWord Tokenization (spaCy):")
for token in doc:
    print(token.text)
```
