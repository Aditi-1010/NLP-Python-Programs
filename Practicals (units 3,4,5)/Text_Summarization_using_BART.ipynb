```python
!pip -q install -U transformers torch

import sys
# Force clear cached transformers modules to resolve the import mismatch error
for module in list(sys.modules.keys()):
    if "transformers" in module:
        sys.modules.pop(module, None)

import torch
from transformers import pipeline

# Using the modern Transformers v5 text-generation pipeline with an instruction model
model_name = "Qwen/Qwen2.5-1.5B-Instruct"

device = 0 if torch.cuda.is_available() else -1

generator = pipeline(
    "text-generation",
    model=model_name,
    device=device,
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32
)

text = """
Artificial intelligence is an important field of computer science.
It enables machines to perform tasks that usually require human
intelligence. Machine learning allows computers to learn patterns
from data. Deep learning uses neural networks to solve complex
problems. AI is used in healthcare, education, transportation,
agriculture, and many other industries. Responsible development
is necessary to ensure that AI systems are safe, fair, and reliable.
"""

messages = [
    {"role": "system", "content": "You are a helpful assistant that summarizes text concisely in 1 to 2 sentences."},
    {"role": "user", "content": f"Please summarize this text:\n{text}"}
]

# Generate the summary
outputs = generator(
    messages,
    max_new_tokens=60,
    do_sample=False
)

print("Original Text:")
print(text)

print("\nGenerated Summary:")
print(outputs[0]["generated_text"][-1]["content"])
```
