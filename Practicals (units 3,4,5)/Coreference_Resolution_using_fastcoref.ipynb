```python
!pip -q install fastcoref
import logging
# Suppress warning messages from huggingface_hub
logging.getLogger("huggingface_hub").setLevel(logging.ERROR)

# Dynamic monkeypatch to fix modern transformers library compatibility
import fastcoref.modeling
if not hasattr(fastcoref.modeling.FCorefModel, "all_tied_weights_keys"):
    fastcoref.modeling.FCorefModel.all_tied_weights_keys = {}

from fastcoref import FCoref

# Initialize the pretrained coreference model
model = FCoref(device="cpu")

text = (
    "Riya is a computer science student. "
    "She loves artificial intelligence. "
    "Riya says she wants to become an AI engineer."
)

# Perform coreference resolution
result = model.predict(texts=[text])

# Display the original text
print("Original Text:")
print(text)

print("\nCoreference Clusters:")

# Display groups of referring expressions
clusters = result[0].get_clusters(as_strings=True)

for cluster in clusters:
    print(cluster)
```
