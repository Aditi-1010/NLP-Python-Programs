```python
import nltk
import gensim.downloader as api
from gensim.models import Word2Vec

nltk.download('punkt')
nltk.download('punkt_tab')

# Sample sentences
sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

# Tokenize
tokens1 = nltk.word_tokenize(sentence1.lower())
tokens2 = nltk.word_tokenize(sentence2.lower())

# Train a small Word2Vec model
sentences = [
    tokens1,
    tokens2,
    ["machine", "learning", "is", "interesting"],
    ["artificial", "intelligence", "is", "interesting"],
    ["I", "love", "technology"],
    ["I", "enjoy", "technology"]
]

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

# Calculate WMD
distance = model.wv.wmdistance(tokens1, tokens2)

print("Sentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("WMD Distance:", distance)

if distance < 1:
    print("Texts are highly similar")
else:
    print("Texts are less similar")
```
